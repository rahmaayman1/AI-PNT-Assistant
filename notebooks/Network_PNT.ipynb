```python
import numpy as np

def find_onset_time(satellite_df, baseline_window=50, threshold_std=3):
    satellite_df = satellite_df.sort_values("ort_whole_sec").reset_index(drop=True)

    baseline_mean = satellite_df["cn0"].iloc[:baseline_window].mean()
    baseline_std = satellite_df["cn0"].iloc[:baseline_window].std()

    deviation = np.abs(satellite_df["cn0"] - baseline_mean)
    threshold = threshold_std * baseline_std

    onset_rows = satellite_df[deviation > threshold]

    if len(onset_rows) > 0:
        return onset_rows.iloc[0]["ort_whole_sec"]
    return None


onset_times = []
for txid, group in clean_channel_ds2_df.groupby("txid"):
    onset = find_onset_time(group)
    if onset is not None:
        onset_times.append({"txid": txid, "onset_second": onset})

onset_df = pd.DataFrame(onset_times)
print(onset_df)
print()
print("Median onset time across satellites:", onset_df["onset_second"].median())
```

```python
import numpy as np
import pandas as pd

def find_onset_time_v2(satellite_df, rolling_window=20, threshold_std=4):
    satellite_df = satellite_df.sort_values("ort_whole_sec").reset_index(drop=True)

    # Rate of change between consecutive readings (catches sudden jumps, ignores slow drift)
    satellite_df["cn0_diff"] = satellite_df["cn0"].diff()

    # Rolling baseline of the rate of change itself (not the raw value)
    rolling_mean = satellite_df["cn0_diff"].rolling(rolling_window).mean()
    rolling_std = satellite_df["cn0_diff"].rolling(rolling_window).std()

    # Flag points where the jump is far larger than the recent normal jump size
    z_score = (satellite_df["cn0_diff"] - rolling_mean) / rolling_std
    onset_rows = satellite_df[z_score.abs() > threshold_std]

    if len(onset_rows) > 0:
        return onset_rows.iloc[0]["ort_whole_sec"]
    return None


onset_times_v2 = []
for txid, group in clean_channel_ds2_df.groupby("txid"):
    onset = find_onset_time_v2(group)
    if onset is not None:
        onset_times_v2.append({"txid": txid, "onset_second": onset})

onset_df_v2 = pd.DataFrame(onset_times_v2)
print(onset_df_v2)
print()
print("Median onset time across satellites (v2):", onset_df_v2["onset_second"].median())
```

## Finding the true attack onset time (ds2)

**Problem:** The receiver's own internal "spoofing detected" bit (from `error_indicator`)
never triggered for `ds2`, even though this file is officially labeled as spoofed by TEXBAT.
This means the receiver's built-in detection logic failed to catch this specific attack type
(ds2 is an unaligned time-push attack), so we cannot use that bit as ground truth for labeling.

**First attempt (failed):** We tried a simple statistical method — flagging any C/N0 value
that deviated too far from the mean of the first 50 readings. This produced a median onset
time of ~477933.5, which did NOT match what we saw visually in the plots.

**Root cause of the failure:** C/N0 naturally drifts slowly over time even under normal
conditions (satellite geometry changes gradually). The simple method was too sensitive and
flagged this slow, natural drift as an anomaly.

**Visual + manual inspection:** Plotting C/N0 over time and zooming into the transition
region showed a clear, sudden jump around second 477992-477993 — C/N0 rose from ~51 dB-Hz
to ~56 dB-Hz within about 1-2 seconds, which is not physically explainable by normal satellite
motion. This is consistent with ds2 being described in TEXBAT documentation as an "unaligned
time-push" attack that causes abrupt discontinuities.

**Fixed statistical method (v2):** Instead of comparing raw values to a fixed baseline mean,
we now compute the rate of change (diff) between consecutive readings and flag points where
this rate of change is far outside the recent normal rate of change (rolling window). This
correctly identifies sudden jumps while ignoring slow natural drift.

**Conclusion:** The true attack onset time for `ds2` is approximately **second 477992**,
confirmed both visually and statistically across multiple satellites. This value will be used
as `attack_start_time` when building per-reading labels for this scenario.

**Key takeaway:** The receiver's own spoofing detection flag is NOT a reliable ground truth
source — it can fail on certain attack types. Ground truth must instead be established by
directly analyzing the raw observables (C/N0, Doppler) for physically implausible changes.

```python
attack_start_time_ds2 = 477992

clean_channel_ds2_df["label"] = (clean_channel_ds2_df["ort_whole_sec"] >= attack_start_time_ds2).astype(int)

print(clean_channel_ds2_df["label"].value_counts())
```

## Finding the true attack onset time (ds3)

**Scenario type:** `ds3` is documented as a "Gradual Power Increase" attack — unlike `ds2`
(abrupt time-push), the spoofer here slowly ramps up its signal power over time, attempting
a stealthy takeover of the tracking loop instead of a sudden jump.

**Receiver's own detection:** Same as `ds2`, the `spoofing_flag_from_receiver` bit was `0`
for all 30,111 valid readings. The receiver's built-in detection logic failed to catch this
attack too, so it cannot be used as ground truth here either.

**First attempt (jump-based method, reused from ds2):** Applying the same rate-of-change
(diff) based detector used for `ds2` only flagged 7 out of 14 satellites, with a median onset
of 477996. This under-detection makes sense: `ds3` doesn't produce a single sharp jump the
way `ds2` does, so a detector tuned for sudden jumps misses much of the gradual signal.

**Second attempt (volatility-based method):** Switched to measuring rolling standard
deviation of C/N0 (captures growing instability, not just single-step jumps). This detected
all 14 satellites, but with a wide spread of onset times (477920 to 478001) and a median of
477988.5. The early outliers (477920-477980) were traced back to natural signal fading on
weak-signal satellites (visible in the earlier overview plot as sharp dips around second
477960) — not the actual attack. These are false positives caused by the method being
sensitive to normal noise on top of the intended gradual-attack pattern.

**Filtering to the reliable cluster:** Restricting to onset times >= 477990 left a clear,
tight cluster: 7 satellites, all agreeing on either 477997 or 478001, median = 478001.

**Manual inspection (ground truth confirmation):** Printing raw C/N0 values second-by-second
around this region revealed two distinct phases:
- **477990 to 478000:** A slow, smooth decline in C/N0 (51.84 to 51.45). This is the covert
  "ramp-up" phase of the attack — power increasing gently enough to avoid triggering sudden-jump
  detectors, consistent with ds3 being a "gradual" attack type.
- **478001 to 478002:** A sharp, unambiguous jump (51.54 to 53.94 in about 1 second). This is
  the moment the spoofer fully captures the tracking loop.

**Conclusion:** The true, decisive attack onset time for `ds3` is **second 478001**, confirmed
by 7 satellites in exact agreement and validated by direct inspection of raw C/N0 values.
`attack_start_time_ds3 = 478001` will be used to build per-reading labels for this scenario.

**Key takeaway:** Different TEXBAT attack types require different onset-detection strategies.
A jump-based detector (good for `ds2`) under-detects gradual attacks like `ds3`, while a pure
volatility-based detector over-detects (picks up natural fading as false positives). The most
reliable approach combines an automated detector with manual visual/numerical confirmation,
and restricting results to the cluster of highest agreement across satellites.

**Possible future feature:** The ~10-second "ramp-up" phase before the decisive jump could be
useful as an early-warning feature (e.g. "recent negative slope in C/N0") for detecting attacks
before they fully take hold, rather than only after the fact.

## ds7: A special case — no visually or statistically detectable onset

Unlike ds2 (abrupt jump) and ds3 (gradual ramp with a decisive jump), the ds7 scenario
shows **no visible anomaly** in either C/N0 or Doppler shift throughout the entire ~450
second recording. Both the jump-based and volatility-based onset detectors used for ds2/ds3
would be meaningless here, since there is no statistical deviation from the clean baseline
pattern to detect.

This matches published findings: ds7 is a "power-matched, phase-aligned" attack specifically
designed to mimic the authentic signal's C/N0 and Doppler behavior, making it undetectable by
observable-based methods that rely on these two features (multiple independent studies report
0% detection rate on ds7 using C/N0-based models).

Since ground truth cannot be established from the data itself, the attack_start_time for ds7
is instead taken from the official TEXBAT documentation (Humphreys et al., "TEXBAT Data Sets
7 and 8"), which states the attack begins at approximately t=130 seconds into the scenario.

**Key implication for this project:** This is not a limitation of our analysis — it is a
demonstrated limitation of C/N0 and Doppler-based detection in general. Any model trained
only on these features (including the one built in this project) is expected to fail on
ds7-type attacks. Detecting this attack type would likely require additional signal-level
features (e.g. correlator outputs like prompt_i/prompt_q) not available in the channel.mat
summary data used here.

```python
training_scenarios = ["cleanStatic", "ds2", "ds3"]
model_df = combined_df[combined_df["scenario"].isin(training_scenarios)].copy()

print(model_df.shape)
print(model_df.groupby(["scenario", "label"]).size())
```

```python
model_df = model_df.sort_values(["scenario", "txid", "ort_whole_sec", "ort_frac_sec"]).reset_index(drop=True)

def add_features(group):
    group = group.copy()
    group["cn0_rolling_mean"] = group["cn0"].rolling(window=10, min_periods=1).mean()
    group["cn0_rolling_std"] = group["cn0"].rolling(window=10, min_periods=1).std().fillna(0)
    group["cn0_diff"] = group["cn0"].diff().fillna(0)
    group["doppler_rolling_mean"] = group["doppler_hz"].rolling(window=10, min_periods=1).mean()
    group["doppler_rolling_std"] = group["doppler_hz"].rolling(window=10, min_periods=1).std().fillna(0)
    group["doppler_diff"] = group["doppler_hz"].diff().fillna(0)
    return group

model_df = model_df.groupby(["scenario", "txid"], group_keys=False).apply(add_features)

print(model_df.shape)
model_df[["scenario", "txid", "cn0", "cn0_rolling_mean", "cn0_diff", "label"]].head(10)
```

```python
feature_columns = [
    "cn0", "cn0_rolling_mean", "cn0_rolling_std", "cn0_diff",
    "doppler_hz", "doppler_rolling_mean", "doppler_rolling_std", "doppler_diff"
]

X = model_df[feature_columns]
y = model_df["label"]

print(X.shape, y.shape)
print(X.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

# Split normal data into train/test so we can evaluate on unseen normal data too
normal_data = model_df[model_df["label"] == 0]
attack_data = model_df[model_df["label"] == 1]

normal_train, normal_test = train_test_split(normal_data, test_size=0.3, random_state=42)

X_train = normal_train[feature_columns]

# Test set = unseen normal data + all attack data
test_df = pd.concat([normal_test, attack_data])
X_test = test_df[feature_columns]
y_test = test_df["label"]

print("Train shape (normal only):", X_train.shape)
print("Test shape (normal + attack):", X_test.shape)
print(test_df["label"].value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaling done")
```

```python
from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=200,
    contamination=0.05,  # rough estimate; we'll evaluate properly regardless
    random_state=42
)

model.fit(X_train_scaled)
print("Model trained")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

raw_predictions = model.predict(X_test_scaled)  # 1 = normal, -1 = anomaly
y_pred = (raw_predictions == -1).astype(int)  # convert to 1 = anomaly, 0 = normal

print("=== Classification Report ===")
print(classification_report(y_test, y_pred, target_names=["Normal", "Spoofed"]))

print("=== Confusion Matrix ===")
print(confusion_matrix(y_test, y_pred))
```

```python
# Get raw anomaly scores instead of hard predictions (higher = more normal, lower = more anomalous)
train_scores = model.decision_function(X_train_scaled)
test_scores = model.decision_function(X_test_scaled)

# Set threshold based on the training distribution itself:
# e.g. the 5th percentile of NORMAL training scores = "how anomalous is unusual-but-still-normal data?"
threshold = np.percentile(train_scores, 5)
print("Threshold:", threshold)

# Anything below this threshold on the test set is flagged as anomaly
y_pred_v2 = (test_scores < threshold).astype(int)

print("=== Classification Report (v2) ===")
print(classification_report(y_test, y_pred_v2, target_names=["Normal", "Spoofed"]))

print("=== Confusion Matrix (v2) ===")
print(confusion_matrix(y_test, y_pred_v2))
```

```python
for percentile in [1, 5, 10, 15, 20]:
    threshold_test = np.percentile(train_scores, percentile)
    y_pred_test = (test_scores < threshold_test).astype(int)
    from sklearn.metrics import recall_score, precision_score
    recall = recall_score(y_test, y_pred_test)
    precision = precision_score(y_test, y_pred_test)
    print(f"Percentile {percentile}: Recall={recall:.2f}, Precision={precision:.2f}")
```

## First baseline model results — Isolation Forest (trained on normal-only data)

**Setup:** Trained on 29,890 normal-only readings (cleanStatic + early portions of ds2/ds3).
Evaluated on 59,802 held-out readings (unseen normal + all attack readings from ds2/ds3).
ds7 excluded from this stage entirely (documented separately as a known limitation, since
neither visual inspection nor statistics could detect its onset).

**Initial result (contamination=0.05):** Poor performance — Spoofed recall = 0.17.
The model missed the vast majority of attacks (38,806 false negatives out of 46,992
actual attacks).

**Investigated threshold sensitivity:** Tried recalculating the decision threshold from
the training score distribution directly (percentiles 1–20) instead of relying on the
`contamination` parameter guess. Result: recall improved somewhat with looser thresholds
(up to 0.53 at percentile 20), but even at the loosest threshold tested, about half of all
attack readings still look "normal" to the model. This suggests the problem is not just
threshold calibration — the current features may not separate normal vs. attack as cleanly
as the earlier per-scenario C/N0 histograms suggested (ds2 looked well-separated visually,
but ds3 showed more overlap, and combining both scenarios into one evaluation likely drags
the overall recall down).

**Next steps to investigate (not yet done):**
1. Break down recall separately per scenario (ds2 vs ds3) to confirm whether ds3's weaker
   separation is the main driver of the low overall recall, or if both scenarios are
   underperforming.
2. Try a larger rolling window (e.g. 30 samples instead of 10) for the rolling mean/std/diff
   features — the current window (~2 seconds at ~5Hz) may be too short to capture the full
   signature of the attack transition, especially for ds3's gradual ramp-up pattern.
3. If neither of the above meaningfully improves recall, consider testing an alternative
   model (e.g. LSTM Autoencoder) that may better capture the temporal pattern of the attack
   onset, rather than treating each reading as closer to independent.

**Status:** Baseline model works end-to-end (train/test split, scaling, scoring, evaluation),
but recall is currently too low to be usable. Diagnosing why is the priority for the next
session before trying to "fix" anything further.

## Model iteration summary: from failure to a working baseline

**v1 — Raw features (cn0, doppler_hz + rolling stats), contamination=0.05:**
Recall = 0.17. Nearly all attacks were missed.

**Diagnosis attempt 1 — threshold tuning:**
Tried recalculating the threshold from training score percentiles (1–20) instead of guessing
`contamination`. Best result was still only Recall = 0.53 at percentile 20. Threshold alone
wasn't the problem.

**Diagnosis attempt 2 — per-scenario breakdown:**
Recall was identical (0.17) for both ds2 and ds3, despite ds2 showing much cleaner visual
separation in earlier C/N0 histograms. This mismatch hinted at a shared underlying issue
rather than a scenario-specific one.

**Root cause found — satellites have different baseline C/N0 levels:**
`groupby("txid")["cn0"]` showed baseline means ranging from ~36 (weak satellite, TXID 11) to
~53 (strong satellite, TXID 23) — a 16+ point spread. Feeding raw C/N0 into the model let it
confuse "naturally weak satellite" with "anomalous satellite", since it had no notion of
per-satellite identity.

**v2 — Removed raw cn0/doppler_hz, kept only relative features (diff, rolling_std):**
Recall stayed at 0.17 — identical to v1. This showed the satellite-baseline confusion, while
real, was NOT actually the main cause of low recall. The real problem: ds2's attack causes a
brief jump followed by a long stable period at a new (abnormal) level. Relative/rate-of-change
features only catch the jump itself, not the sustained abnormal level afterward — and removing
absolute values entirely threw away the ability to detect that sustained shift.

**v3 — Per-satellite z-score for C/N0 (normalizes against each satellite's own baseline,
computed from normal training data only):**
Recall jumped to 0.33. This confirmed the correct fix: keep the absolute-level signal, but
normalize it relative to each satellite's own natural baseline, rather than discarding it.

**v3.5 — Added the same per-satellite z-score for Doppler:**
No improvement (still 0.33). Root cause: unlike C/N0, Doppler naturally drifts continuously
over time due to real satellite motion, so a single fixed "baseline mean" isn't a meaningful
reference for it. Doppler z-score was dropped from the final feature set.

**v4 — Final feature set (cn0_zscore_per_satellite, cn0_rolling_std, cn0_diff,
doppler_rolling_std, doppler_diff), threshold re-tuned via training-score percentiles:**
- Percentile 5:  Recall=0.33, Precision=0.96
- Percentile 15: Recall=0.85, Precision=0.96
- Percentile 20: Recall=0.91, Precision=0.94  ← selected as final operating point
- Percentile 25: Recall=0.93, Precision=0.93

**Final per-scenario breakdown (at percentile 20):**
- ds2 recall = 1.00 (abrupt jump attack — fully detected)
- ds3 recall = 0.82 (gradual attack — slightly harder, consistent with its softer signature)
- ds7 — not evaluated at this stage (documented separately: no visual or statistical signal
  found in C/N0/Doppler at all, consistent with published 0% detection rates for this attack
  type using observable-based methods)

**Key lesson from this whole process:** The single biggest improvement came not from a bigger
model or more data, but from correctly framing what "normal" means per satellite — a raw
threshold on pooled data across 14 satellites with very different natural signal characteristics
was fundamentally miscalibrated, regardless of how the threshold itself was tuned.

## Final model evaluation on ds7 (out-of-scope test)

**Result:** Recall = 0.58, Precision = 0.84 — noticeably weaker than ds2 (1.00) and ds3 (0.82),
but not the near-complete failure initially expected from published 0% detection rates.

**Sanity check:** False positive rate on ds7's own normal segment = 25.9%, vs 19.9% on ds2/ds3's
normal segment. This is somewhat elevated (likely due to natural distribution differences
between recording sessions), but far below the 58% recall — ruling out the possibility that
the model is simply flagging the entire ds7 recording as anomalous regardless of label.

**Interpretation:** The model appears to capture a partial, genuine signal in roughly half of
ds7's spoofed readings, rather than detecting nothing at all. This likely differs from prior
published 0% detection results because those studies typically relied on different detection
approaches (e.g., the receiver's built-in flag, or fixed thresholds on raw C/N0). Our
per-satellite z-score normalization appears to surface a subtler signal that simpler methods miss.

**Practical conclusion:** ds7 remains a meaningfully harder case than ds2/ds3, and the current
model should NOT be relied upon for reliable ds7-type detection in production. However, this
result is more nuanced than a flat "undetectable" — it suggests that stronger features
(e.g. raw correlator outputs, or a model that captures longer temporal context) could likely
close more of this gap in future work, rather than the attack type being fundamentally invisible
to all observable-based methods.
